```python
#Traduccion a
a = 1
b = 1
n=10
for i in range(3, n):
    aux = a + b
    a = b
    b = aux

print(b)
```

```python
#Traduccion B
a = 1
b = 1
n=10
for i in range(3, n + 1):
    a, b = b, a + b

print(b)
```

```python
#Punto4
n = int(input())

es_primo = n >= 2
d = 2

while d * d <= n and es_primo:
    if n % d == 0:
        es_primo = False

    d = d + 1

print(es_primo)
```
